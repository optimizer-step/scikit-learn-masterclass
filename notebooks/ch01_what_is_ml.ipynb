```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size = 0.25,            # Hides the 25% of the dataset from the model
    random_state = 42,
    stratify = y,

)

print("train: ", X_train.shape, "test:", X_test.shape)
print("failure rate train:", round(y_train.mean(), 3), "test:", round(y_test.mean(), 3))
```

```python
rates = []
for seed in [0,1,2,3,4]:
    _, _, _, y_te = train_test_split(X, y, test_size = 0.25, random_state = seed)
    rates.append(y_te.mean())
    print(f"seed {seed} test failure rate = {y_te.mean():.3f}")
```

```python
from sklearn.dummy import DummyClassifier

baseline = DummyClassifier(strategy = 'most_frequent')
baseline.fit(X_train, y_train)
print(baseline.predict(X_test)[:100])
```

```python
y_train.value_counts()
```

```python
baseline_acc = baseline.score(X_test, y_test)
print("Baseline accuracy", round(baseline_acc, 3))
```

```python
df['failed'].value_counts()
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model = LogisticRegression(max_iter = 1000)                   ## Classification model

print("learned anything yet ?", hasattr(model, "coef_"))

model.fit(X_train[['grid_load_index']], y_train)
print("Has model learned anything now ?", hasattr(model, "coef_"))
```

```python
model.coef_
```

```python
print("Coefficient:", model.coef_)
print("intercept:", model.intercept_)
print("Classes", model.classes_)
print()
print("Parameters we could have set: C = ", model.C,
"max_iter = ", model.max_iter, " solver =", model.solver)
```

```python
y_train.value_counts()
```

```python
X_test_one = X_test[['grid_load_index']]

labels = model.predict(X_test_one)
probs = model.predict_proba(X_test_one)

print("labels :", labels[:8])
print("P(fail):", probs[:8].round(3))
print()
print("highest P(fail) in the whole test set:", round(probs.max(), 4))
```

```python
acc = model.score(X_test_one, y_test)
print("LogisticRegression model accuracy: ", round(acc, 4))
print("Baseline accuracy: ", round(baseline_acc, 4))
```
